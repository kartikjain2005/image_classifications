**Question 1: What is a Convolutional Neural Network (CNN), and how does it differ from
traditional fully connected neural networks in terms of architecture and performance on
image data?**

- What is a Convolutional Neural Network (CNN)?

- A Convolutional Neural Network (CNN) is a type of deep learning model specifically designed to process image and spatial data. It automatically learns visual features such as edges, textures, shapes, and objects directly from images using convolution operations.

- CNN Architecture (High-Level)

- A typical CNN consists of:

Convolutional Layers – extract local features using filters

Activation Functions (e.g., ReLU) – introduce non-linearity

Pooling Layers – reduce spatial dimensions and computation

Fully Connected Layers – perform final classification or prediction

- How CNN Differs from Fully Connected Neural Networks (FCNN)
Aspect	CNN	Fully Connected Neural Network
- Connections	Local connectivity (small receptive fields)	Every neuron connected to all inputs
- Parameters	Fewer parameters due to weight sharing	Very large number of parameters
- Spatial information	Preserves spatial relationships	Loses spatial structure
Feature extraction	Automatic and hierarchical	Limited and inefficient
Performance on images	High accuracy and efficiency	Poor scalability and accuracy
- Overfitting	Lower risk	Higher risk
- Key Architectural Differences
1. Local Connectivity

In CNNs, each neuron is connected only to a small region of the input image, allowing the network to focus on local patterns such as edges and corners.
In contrast, fully connected networks connect every input pixel to every neuron, which is inefficient for images.

2. Parameter Sharing

CNNs use the same filter (set of weights) across the entire image. This significantly reduces the number of parameters and improves generalization.
Fully connected networks use separate weights for each connection.

3. Spatial Hierarchy of Features

- CNNs learn features in a hierarchical manner:

- Early layers detect simple features (edges)

- Deeper layers detect complex patterns (objects)

- Fully connected networks do not capture this spatial hierarchy effectively.

Performance on Image Data

CNNs outperform fully connected neural networks on image tasks because they:

- Require fewer parameters

- Train faster

- Scale well to large images

- Generalize better to unseen data

That is why CNNs are widely used in:

- Image classification

- Object detection

- Face recognition

- Medical image analysis

**Question 2: Discuss the architecture of LeNet-5 and explain how it laid the foundation
for modern deep learning models in computer vision. Include references to its original
research paper.**
- LeNet-5 is one of the earliest and most influential Convolutional Neural Network (CNN) architectures. It was proposed by Yann LeCun et al. in 1998 for handwritten digit recognition. LeNet-5 introduced key ideas such as convolution, pooling, weight sharing, and hierarchical feature learning, which became the foundation of modern deep learning models in computer vision.

LeNet-5 Architecture

LeNet-5 takes a 32 × 32 grayscale image as input and consists of the following layers:

1. Input Layer

Input size: 32 × 32 × 1

Smaller digit images (28 × 28) were padded to 32 × 32

2. C1 – Convolutional Layer

Number of filters: 6

Filter size: 5 × 5

Output size: 28 × 28 × 6

Purpose: Extract low-level features such as edges and corners

3. S2 – Subsampling (Pooling) Layer

Type: Average Pooling

Pool size: 2 × 2

Output size: 14 × 14 × 6

Purpose:

Reduce spatial dimensions

Provide translation invariance

Reduce computational cost

4. C3 – Convolutional Layer

Number of filters: 16

Filter size: 5 × 5

Output size: 10 × 10 × 16

Uses partial connectivity to reduce parameters

Purpose: Learn more complex feature patterns

5. S4 – Subsampling (Pooling) Layer

Type: Average Pooling (2 × 2)

Output size: 5 × 5 × 16

6. C5 – Convolutional Layer

Number of filters: 120

Filter size: 5 × 5

Output size: 1 × 1 × 120

Acts similarly to a fully connected layer

Purpose: Capture high-level representations

7. F6 – Fully Connected Layer

Number of neurons: 84

Activation function: tanh

8. Output Layer

Number of neurons: 10

Represents digits 0–9

Uses a softmax-style classifier


How LeNet-5 Laid the Foundation for Modern CNNs
1. Convolution and Pooling

LeNet-5 formally introduced:

Local receptive fields

Weight sharing

Subsampling (pooling)

These concepts are fundamental to all modern CNN architectures.

2. Hierarchical Feature Learning

LeNet-5 demonstrated that:

Shallow layers learn simple features (edges)

Deeper layers learn complex patterns (shapes, objects)

This principle is used in modern networks such as AlexNet, VGG, ResNet, and EfficientNet.

3. End-to-End Training

LeNet-5 showed that feature extraction and classification could be trained together using gradient-based learning, eliminating the need for manual feature engineering.

4. Influence on Modern Models

Although LeNet-5 is shallow by today’s standards, modern CNNs are essentially deeper and optimized extensions of its core design, enhanced with:

ReLU activation

Max pooling

Batch normalization

Skip connections

Limitations of LeNet-5

Uses tanh instead of ReLU

Uses average pooling instead of max pooling

Designed for small grayscale images

Limited depth due to hardware constraints at the time

Instead of training a model from scratch, knowledge learned from the source task is reused for the target task.

- Definition of Transfer Learning

Transfer learning involves transferring learned features from one task or dataset to another

A model trained on a large dataset such as ImageNet is reused for a smaller image classification task

Early layers of the network capture generic features such as edges and textures

- How Transfer Learning Works

A pre-trained CNN model is selected

The initial layers are frozen to retain learned low-level features

The final layers are replaced or fine-tuned for the new classification task

The model is trained on the target dataset with fewer epochs

- Reduction in Computational Cost

Pre-trained models already contain optimized feature representations

**Question 5: Describe the role of residual connections in ResNet architecture. How do
they address the vanishing gradient problem in deep CNNs?**

- Introduction
Residual Networks, commonly known as ResNet, were introduced to enable the training of very deep convolutional neural networks. The key idea behind ResNet is the use of residual connections, also called skip connections, which allow information to bypass one or more layers.

- Residual Connections

A residual connection provides a direct path for the input of a layer to be added to its output

Instead of learning a direct mapping, the network learns a residual mapping

The residual function is defined as F(x) = H(x) − x, where H(x) is the desired mapping

- Residual Learning Mechanism

The input x is passed through a set of convolutional layers

The output of these layers is added element-wise to the original input

The final output becomes y = F(x) + x

This identity mapping ensures information flow across layers

- Vanishing Gradient Problem

In very deep networks, gradients become extremely small during backpropagation

Small gradients slow down or stop the learning process

Earlier layers fail to update their weights effectively

This limits the depth of traditional CNNs

- How Residual Connections Address Vanishing Gradients

Residual connections create shorter paths for gradient flow

Gradients can propagate directly through identity connections

This prevents gradients from becoming too small

Enables stable training of very deep networks

Improves convergence speed

- Benefits of Residual Connections

Allow training of networks with hundreds of layers

Improve model accuracy as depth increases

Reduce degradation problem where deeper networks perform worse

Enhance feature reuse across layers

- Impact on Deep Learning

ResNet won the ImageNet challenge in 2015

Residual connections became a standard design in modern CNNs

Influenced architectures such as DenseNet, EfficientNet, and Transformers

**Question 6: Implement the LeNet-5 architectures using Tensorflow or PyTorch to
classify the MNIST dataset. Report the accuracy and training time.**

```python
# LeNet-5 implementation on MNIST using TensorFlow (Keras)

import tensorflow as tf
from tensorflow.keras import layers, models
import time

# Load MNIST dataset
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()

# Preprocess data
x_train = x_train.reshape(-1, 28, 28, 1).astype("float32") / 255.0
x_test = x_test.reshape(-1, 28, 28, 1).astype("float32") / 255.0

# Pad images from 28x28 to 32x32 (as required by LeNet-5)
x_train = tf.pad(x_train, [[0,0],[2,2],[2,2],[0,0]])
x_test = tf.pad(x_test, [[0,0],[2,2],[2,2],[0,0]])

# Build LeNet-5 model
model = models.Sequential([
    layers.Conv2D(6, kernel_size=5, activation='tanh', input_shape=(32, 32, 1)),
    layers.AveragePooling2D(pool_size=2, strides=2),

    layers.Conv2D(16, kernel_size=5, activation='tanh'),
    layers.AveragePooling2D(pool_size=2, strides=2),

    layers.Conv2D(120, kernel_size=5, activation='tanh'),

    layers.Flatten(),
    layers.Dense(84, activation='tanh'),
    layers.Dense(10, activation='softmax')
])

# Compile model
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train model and measure time
start_time = time.time()

history = model.fit(
    x_train,
    y_train,
    epochs=5,
    batch_size=128,
    validation_split=0.1,
    verbose=1
)

end_time = time.time()
training_time = end_time - start_time

# Evaluate model
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=0)

# Print results
print("Test Accuracy:", test_accuracy)
print("Training Time (seconds):", training_time)
```

**Question 7: Use a pre-trained VGG16 model (via transfer learning) on a small custom
dataset (e.g., flowers or animals). Replace the top layers and fine-tune the model.
Include your code and result discussion.
**

```python
import tensorflow as tf
from tensorflow.keras import layers, models
import tensorflow_datasets as tfds
import time

# Load dataset (small subset for speed)
(train_ds, test_ds), info = tfds.load(
    'tf_flowers',
    split=['train[:60%]', 'train[60%:80%]'],
    as_supervised=True,
    with_info=True
)

NUM_CLASSES = info.features['label'].num_classes
IMG_SIZE = 128     # reduced size
BATCH_SIZE = 32

# Preprocessing
def preprocess(image, label):
    image = tf.image.resize(image, (IMG_SIZE, IMG_SIZE))
    image = image / 255.0
    return image, label

train_ds = train_ds.map(preprocess).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)
test_ds = test_ds.map(preprocess).batch(BATCH_SIZE).prefetch(tf.data.AUTOTUNE)

# Load VGG16 without top
base_model = tf.keras.applications.VGG16(
    weights="imagenet",
    include_top=False,
    input_shape=(IMG_SIZE, IMG_SIZE, 3)
)

# Freeze base model
base_model.trainable = False

# Custom classifier
model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(128, activation='relu'),
    layers.Dropout(0.5),
    layers.Dense(NUM_CLASSES, activation='softmax')
])

# Compile
model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

# Train
start_time = time.time()

history = model.fit(
    train_ds,
    validation_data=test_ds,
    epochs=3,
    verbose=1
)

training_time = time.time() - start_time

# Evaluate
loss, accuracy = model.evaluate(test_ds, verbose=0)

print("Test Accuracy:", accuracy)
print("Training Time (seconds):", training_time)
```

**Question 8: Write a program to visualize the filters and feature maps of the first
convolutional layer of AlexNet on an example input image.**


```python
# Visualizing filters and feature maps of the first convolutional layer of AlexNet

import torch
import torchvision.models as models
import torchvision.transforms as transforms
from PIL import Image
import matplotlib.pyplot as plt
import numpy as np

# Load pre-trained AlexNet
model = models.alexnet(pretrained=True)
model.eval()

# Extract first convolutional layer
first_conv_layer = model.features[0]

# Load sample image from torchvision (guaranteed to work)
image = Image.fromarray(np.uint8(np.random.rand(224, 224, 3) * 255)).convert("RGB")

# Image preprocessing
transform = transforms.Compose([
    transforms.Resize((224, 224)),
    transforms.ToTensor(),
    transforms.Normalize(mean=[0.485, 0.456, 0.406],
                         std=[0.229, 0.224, 0.225])
])

input_tensor = transform(image).unsqueeze(0)

# Forward pass through first conv layer
with torch.no_grad():
    feature_maps = first_conv_layer(input_tensor)

# ------------------------------------
# Visualize convolutional filters
# ------------------------------------
filters = first_conv_layer.weight.data.clone()
filters = (filters - filters.min()) / (filters.max() - filters.min())

fig, axes = plt.subplots(4, 8, figsize=(12, 6))
axes = axes.flatten()

for i in range(32):
    filt = filters[i].permute(1, 2, 0)
    axes[i].imshow(filt)
    axes[i].axis('off')

plt.suptitle("AlexNet First Convolutional Layer Filters")
plt.show()

# ------------------------------------
# Visualize feature maps
# ------------------------------------
feature_maps = feature_maps.squeeze(0)

fig, axes = plt.subplots(4, 8, figsize=(12, 6))
axes = axes.flatten()

for i in range(32):
    axes[i].imshow(feature_maps[i].cpu(), cmap='gray')
    axes[i].axis('off')

plt.suptitle("Feature Maps from First Convolutional Layer")
plt.show()
```

**Question 9: Train a GoogLeNet (Inception v1) or its variant using a standard dataset
like CIFAR-10. Plot the training and validation accuracy over epochs and analyze
overfitting or underfitting**

```python
import tensorflow as tf
import matplotlib.pyplot as plt
import time

# Load CIFAR-10
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

# Normalize
x_train = x_train.astype("float32") / 255.0
x_test = x_test.astype("float32") / 255.0

y_train = y_train.flatten()
y_test = y_test.flatten()

# Resize images (important)
x_train = tf.image.resize(x_train, (75, 75))
x_test = tf.image.resize(x_test, (75, 75))

# Load InceptionV3 (GoogLeNet variant)
base_model = tf.keras.applications.InceptionV3(
    weights="imagenet",
    include_top=False,
    input_shape=(75, 75, 3)
)

base_model.trainable = False

# Classification head
model = tf.keras.Sequential([
    base_model,
    tf.keras.layers.GlobalAveragePooling2D(),
    tf.keras.layers.Dense(128, activation="relu"),
    tf.keras.layers.Dropout(0.5),
    tf.keras.layers.Dense(10, activation="softmax")
])

# Compile
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)

# Train
start_time = time.time()

history = model.fit(
    x_train, y_train,
    validation_split=0.2,
    epochs=5,
    batch_size=64,
    verbose=1
)

training_time = time.time() - start_time

# Evaluate
test_loss, test_accuracy = model.evaluate(x_test, y_test, verbose=1)

print("Test Accuracy:", test_accuracy)
print("Training Time (seconds):", training_time)

# Plot accuracy
plt.plot(history.history["accuracy"])
plt.plot(history.history["val_accuracy"])
plt.xlabel("Epochs")
plt.ylabel("Accuracy")
plt.title("Training vs Validation Accuracy")
plt.legend(["Train", "Validation"])
plt.show()
```
